# Data Cleaning & Preprocessing

This notebook performs data preprocessing on the raw dataset and exports a cleaned CSV file for further stages (EDA & Feature Engineering).

## 1. Import Libraries

```

import pandas as pd
import numpy as np
```

## 2. Load Dataset

```

# Update the path if needed
df = pd.read_csv("raw_data.csv")
df.head()
```

## 3. Initial Data Inspection

```

df.info()
df.describe()
```

## 4. Data Type Conversion

```

# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Convert Wake_Up_Time to datetime and extract hour
df['Wake_Up_Time'] = pd.to_datetime(df['Wake_Up_Time'], format='%H:%M')
df['Wake_Up_Hour'] = df['Wake_Up_Time'].dt.hour

df.drop(columns=['Wake_Up_Time'], inplace=True)
```

## 5. Handle Identifier Column

```

# User_ID is an identifier, drop it
df.drop(columns=['User_ID'], inplace=True)
```

## 6. Outlier Treatment using IQR

```

numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    df[col] = np.clip(df[col], lower, upper)
```

## 7. Final Check

```

df.info()
df.isnull().sum()
```

## 8. Export Cleaned Dataset

```

df.to_csv("cleaned_data.csv", index=False)
print("cleaned_data.csv exported successfully")
```
