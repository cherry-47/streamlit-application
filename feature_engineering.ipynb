# Feature Engineering

This notebook performs advanced feature engineering to enhance classification performance.
The engineered dataset will be exported for model training.

## 1. Import Required Libraries

```

import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

## 2. Load Cleaned Dataset

```

df = pd.read_csv("data/cleaned_data.csv")
df.head()
```

## 3. Sleep Category Feature

```

def sleep_category(hours):
    if hours < 6:
        return "Low Sleep"
    elif hours <= 8:
        return "Optimal Sleep"
    else:
        return "High Sleep"

df["Sleep_Category"] = df["Sleep_Hours"].apply(sleep_category)
```

## 4. Lifestyle-Based Engineered Features

```

df["Activity_Score"] = df["Steps"] + df["Calories_Burned"]
df["Hydration_Liters"] = df["Water_Intake_ml"] / 1000
df["Study_Effort"] = df["Study_Hours"] * 10
```

## 5. Lifestyle Balance Index

```

df["Lifestyle_Balance_Index"] = (
    df["Sleep_Hours"] * 0.30 +
    df["Hydration_Liters"] * 0.20 +
    df["Study_Hours"] * 0.30 +
    df["Activity_Score"] * 0.00005
)
```

## 6. Encode Categorical Features

```

df = pd.get_dummies(df, columns=["Sleep_Category"], drop_first=True)
```

## 7. Feature Scaling

```

scaler = StandardScaler()
num_cols = df.drop(columns=["Mood_Score"]).select_dtypes(include=["int64", "float64"]).columns
df[num_cols] = scaler.fit_transform(df[num_cols])
```

## 8. Final Dataset Preview

```

df.head()
```

## 9. Export Engineered Dataset

```

df.to_csv("data/engineered_data.csv", index=False)
print("Engineered dataset saved as data/engineered_data.csv")
```
