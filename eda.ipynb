# Exploratory Data Analysis (EDA)

This notebook performs **statistical and visualization-based exploration** on the cleaned dataset to understand patterns, distributions, and relationships with the target variable **Mood_Score**.

## 1. Import Required Libraries

```

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

## 2. Load Cleaned Dataset

```

df = pd.read_csv("cleaned_data.csv")
df.head()
```

## 3. Dataset Overview

```

df.info()
df.describe().T
```

## 4. Target Variable Distribution (Mood_Score)

```

plt.figure(figsize=(10,5))
sns.countplot(x='Mood_Score', data=df, palette='viridis')
plt.title("Distribution of Mood Score (Target Variable)")
plt.show()
```

## 5. Numerical Feature Distributions

```

num_cols = df.select_dtypes(include=['int64','float64']).columns

df[num_cols].hist(figsize=(16,12), bins=20)
plt.suptitle("Distribution of Numerical Features")
plt.show()
```

## 6. Boxplots for Outlier Verification

```

plt.figure(figsize=(16,10))
for i, col in enumerate(num_cols, 1):
    plt.subplot(3, 3, i)
    sns.boxplot(y=df[col], color='skyblue')
    plt.title(col)
plt.tight_layout()
plt.show()
```

## 7. Correlation Analysis

```

corr = df[num_cols].corr()

plt.figure(figsize=(12,8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Heatmap")
plt.show()
```

## 8. Pairplot (Key Relationships)

```

selected_features = ['Sleep_Hours', 'Steps', 'Calories_Burned', 
                     'Water_Intake_ml', 'Study_Hours', 'Mood_Score']

sns.pairplot(df[selected_features], hue='Mood_Score', palette='tab10')
plt.show()
```

## 9. Feature vs Target Analysis

```

plt.figure(figsize=(14,8))
sns.boxplot(x='Mood_Score', y='Sleep_Hours', data=df)
plt.title("Sleep Hours vs Mood Score")
plt.show()

plt.figure(figsize=(14,8))
sns.boxplot(x='Mood_Score', y='Steps', data=df)
plt.title("Steps vs Mood Score")
plt.show()

plt.figure(figsize=(14,8))
sns.boxplot(x='Mood_Score', y='Study_Hours', data=df)
plt.title("Study Hours vs Mood Score")
plt.show()
```

## 10. Insights Summary

- Mood_Score is reasonably well distributed across classes.
- Lifestyle features such as Sleep_Hours, Steps, and Study_Hours show visible relationships with Mood_Score.
- No severe multicollinearity detected.
- Dataset is suitable for classification modeling.
